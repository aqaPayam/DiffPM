```python
import numpy as np
import torch
import seaborn as sns
import matplotlib.pyplot as plt
import os

# 1) Prepare the normalized log‐return series from the penguins dataset
print("→ Loading penguins dataset from seaborn…")
df = sns.load_dataset("penguins")
FEATURE = "body_mass_g"
raw = df[FEATURE].dropna().to_numpy(dtype=np.float32)
log_returns = np.log(raw[1:] / raw[:-1])
series = (log_returns - log_returns.mean()) / log_returns.std()
series = raw

print(f"✔️  Normalized log‐returns: mean {series.mean():.2f}, std {series.std():.2f}")

# 2) Reshape into (N, T, D) = (1, len(series), 1) and save to .npy
arr = series[np.newaxis, :, np.newaxis]   # shape (1, T, 1)
os.makedirs("data", exist_ok=True)
np.save("data/penguins_series.npy", arr)

# 3) Import and run the diffusion training
from run import run  # adjust import path as needed

device = torch.device("cuda" if torch.cuda.is_available() else "cpu")

model = run(
    data_npy="data/penguins_series.npy",
    ma_window_size=2,       # moving‐average window for trend decomposition
    window_size=10,         # model window size W
    time_emb_dim=128,       # default time embedding dimension
    base_channels=64,       # default U‑Net base channels
    n_res_blocks=4,         # default number of ResBlocks
    timesteps=1000,         # T = 1000 diffusion steps
    s=0.007,                # cosine schedule offset
    batch_size=64,          # BATCH_SIZE = 64
    epochs=500,             # EPOCHS = 500
    lr=1e-3,                # LR = 1e-3
    device=device,
    show_detail=True,       # plot samples every sample_interval
    sample_interval=5       # now plotting every 5 epochs
)

print("✅ Training complete. Returned model:", model)
```
